```python
# 10.Concatenate and Merge Dataframe (Joins) 
```

```python
df1 = pd.DataFrame({'ID':[1,2,3,4],'Name':['A','B','C','D']})
df1
```

```python
df2 = pd.DataFrame({'ID':[1,2,3,4],'Score':[88,98,76,94]})
df2
```

```python
pd.concat([df1,df2],axis=0) # vertical / row level / top on
```

```python
pd.concat([df1,df2],axis=1) # horizantal / column level / side on side
```

```python
# Merge = join
pd.merge(df1,df2,how = 'inner',on = 'ID')
```

```python
pd.merge(df1,df2,how = 'left',on = 'ID')
```

```python
pd.merge(df1,df2,how = 'right',on = 'ID')
```

```python
pd.merge(df1,df2,how = 'outer',on = 'ID')
```

```python
pd.merge(df1,df2,how = 'inner',left_on = 'ID',right_on='ID')
```
